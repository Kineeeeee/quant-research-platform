# Research summary

Index of what the four research notebooks actually found, so I don't have to
re-open each one to remember. Short version: most ideas were weak or didn't
work, which is the normal result and the whole point of testing first.

## NB01 - Stock momentum
Do past winners keep winning? Signal: 6M past return (skip the last month),
long top 20% / short bottom 20%, hold 1 month.

Real but modest. On non-overlapping monthly spreads: Sharpe ~0.67, p=0.042
(just significant). Max drawdown ~-43%. The first run looked amazing (t~7,
Sharpe 2.3) but that was overlapping-returns inflation, not alpha - once I
subsampled to independent windows it came back to earth. Worth trading with a
crash/vol filter, not as a free lunch.

## NB02 - Mean reversion
Do oversold stocks bounce? Signal: 5-21 day past return, bottom 20% = oversold.

No. Every combo is insignificant and the Sharpes are slightly negative. On
large-cap US names the losers kept losing, which is really short-term momentum
again, not reversion. Adding an RSI filter didn't help. Don't build short-term
MR here; it'd need small-caps or intraday.

## NB03 - Volatility clustering
Do big moves cluster, and is vol predictable?

Yes, clearly. Raw returns have ~zero autocorrelation (direction is
unpredictable), but absolute returns are strongly autocorrelated (~0.32 at
lag 1, still ~0.24 at lag 20) - the GARCH effect. An EWMA vol forecast
correlates ~0.68 with next-month realised vol (R^2 ~0.46). Fat tails
everywhere. This is the one strong, usable result: size positions off vol,
since vol is the forecastable part. Feeds directly into the vol-adjusted
momentum strategy and the risk manager.

## NB04 - Sector rotation
Is 3-month sector relative momentum tradeable? Long top 3 / short bottom 3
sector ETFs.

No, at 3M it's noise (t~-0.15, p~0.9). Sectors are too correlated for
short-term momentum to produce a real spread. Multi-year trends are obvious in
hindsight (Tech up, Energy down then back), but a tradeable 3M signal isn't
there - would need a longer lookback or a regime trigger.

## Takeaways
- Momentum: weak but real, trade it with a crash filter.
- Short-term mean reversion: not on this universe.
- Volatility clustering: strong and the most useful finding - drives sizing.
- Sector rotation (3M): noise.

Three of four hypotheses were basically negatives. That's expected, and the
value is in not having traded them.
